```python
image = cv2.imread("image.png", cv2.IMREAD_COLOR)

if image is None:
    print("Pas bon")
```

```python
import matplotlib.pyplot as plt

print(image.shape)

plt.imshow(image)
plt.show()
```

```python
import numpy as np

gray = np.zeros(image.shape[:2])

for x, row in enumerate(image):
    for y, pix in enumerate(row):
        gray[x, y] = np.sum(pix) / 3
```
